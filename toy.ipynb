```python
import numpy as np
from scipy.sparse import csr_matrix

A = csr_matrix(np.random.randn(8, 8))
# A_bsr = A.tobsr(blocksize=(2, 2))
A_bsr = A.tobsr()

print(A_bsr.blocksize)   # (2, 2)
print(A_bsr.data.shape) # (n_blocks, 2, 2)

import numpy as np
from scipy.sparse import block_diag

# Three dense blocks (2x2 each)
B1 = np.array([[1., 2.],
               [3., 4.]])
B2 = np.array([[5., 0.],
               [0., 6.]])
B3 = np.array([[7., 8.],
               [9., 10.]])

# Build sparse block-diagonal matrix (CSR by default)
A_csr = block_diag((B1, B2, B3), format="csr")

print(A_csr.shape)
# (6, 6)

# A_bsr = A_csr.tobsr(blocksize=(2, 2))
A_bsr = A_csr.tobsr()
print("Block size:", A_bsr.blocksize)
print("Number of blocks:", A_bsr.data.shape[0])
print("BSR data:")
print(A_bsr.data)
```
